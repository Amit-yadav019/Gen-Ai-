```python
# Cell 1 — Verify everything works
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
result = classifier("I am excited to learn HuggingFace!")
print(result)
```

```python
# Cell 2 — Text generation (SAFE for 8GB Mac)
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")
result = generator(
    "In this course, we will teach you how to",
    max_length=50,
    truncation=True
)
print(result[0]['generated_text'])
```

```python
# Cell 3 — Question Answering
from transformers import pipeline

qa = pipeline("question-answering")
result = qa(
    question="Where do I study?",
    context="My name is Amit and I study at NIT Jamshedpur"
)
print(result)
# Output: {'answer': 'NIT Jamshedpur', 'score': 0.99...}
```
